```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

qualifying = pd.read_csv('Qualifying_times.csv').copy()
race = pd.read_csv('Race_results.csv').copy()

"""
What to do with the data

2 speed to race dominance comparision

3 storytelling of what was happening

"""


qualifying['Pos'] = qualifying.index
race['Pos'] = race.index
qualifying.set_index("Driver", inplace=True)
race.set_index("Driver", inplace=True)
qualifying["Movement"] = qualifying["Pos"] - race["Pos"]
diff_plot = qualifying.reset_index()
plt.figure(figsize=(10, 6))
plt.scatter(
    diff_plot["Pos"], diff_plot["Movement"], color="blue", alpha=0.7, s=100
)
for i, txt in enumerate(diff_plot["Driver"]):
    plt.annotate(
        txt,
        (diff_plot["Pos"].iloc[i], diff_plot["Movement"].iloc[i]),
        textcoords="offset points",
        xytext=(5, 5),
        ha="left",
        fontsize=9,
    )
plt.axhline(0, color="red", linestyle="--", linewidth=1)  # Line showing 0 movement
plt.title("Driver Qualifying Position vs Positions gained/lost on Race day", fontsize=14)
plt.xlabel("Original Position in Qualifying", fontsize=12)
plt.ylabel("Places Moved Up / Down", fontsize=12)
plt.grid(True, linestyle=":", alpha=0.6)

plt.tight_layout()
plt.show()



```

This is the first chart I made for this Race analysis. I compared the qualifying race position and compared it with how many places each driver gained/lost, which tells a story. Before going further, it is important to note that this analysis is under 30 points, so the distributions can be incredibly skewed, hence you need to look at this at a sports perspective rather than looking at it face value and purely from a statistical way.

Drivers on the zero line show that they drove consistiently, their qualifying pace matched on their race pace. The four drivers on this line really performed as expected. Lando Norris had a sensational race, given that the Mclaren team were strugling in the first two rounds this year. Everyone else on the line had a difficult race, as these results aren't ones to remember.


However, anyone who was above the line indicate that they improved from their qualifying pace, either due to race pace improving or strategy calls. Some key drivers in this section were the Red bulls, who are having some solid consistency amid issues with the battery and wings, the Aston Martins and their new B-spec car improving the pace but still in the back of the pack, and Nico Hulkenberg, who was able to extract and convert the Audi's speed for some well needed points.

The ones under the line tell a different story, as some like Charles Leclerc, Arvid Lindblad, and Esteban Ocon did pretty well but couldn't retain their qualifying pace. Lewis Hamilton suffered pretty badly too, losing 3 positions. But, the outlier of this dataset is Oscar Piastri, due to mechanical failures caused on-track collisions.


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

fast_laps = pd.read_csv("Fastest_laps.csv").copy()
race = pd.read_csv('Race_results.csv').copy()
fast_laps['Pos'] = fast_laps.index
race['Pos'] = race.index
fast_laps.set_index("Driver", inplace=True)
race.set_index("Driver", inplace=True)
race["Movement"] = fast_laps["Pos"] - race["Pos"]
diff_plot = race.reset_index()
plt.figure(figsize=(10, 6))
plt.scatter(
    diff_plot["Pos"], diff_plot["Movement"], color="blue", alpha=0.7, s=100
)
for i, txt in enumerate(diff_plot["Driver"]):
    plt.annotate(
        txt,
        (diff_plot["Pos"].iloc[i], diff_plot["Movement"].iloc[i]),
        textcoords="offset points",
        xytext=(5, 5),
        ha="left",
        fontsize=6,
    )
plt.axhline(0, color="red", linestyle="--", linewidth=1)  # Line showing 0 movement
plt.title("Driver Race Day vs Positions gained/lost in fast laps", fontsize=14)
plt.xlabel("Original Position on Race day", fontsize=12)
plt.ylabel("Places Moved Up / Down", fontsize=12)
plt.grid(True, linestyle=":", alpha=0.6)

plt.tight_layout()
plt.show()
```

Although this graph seems similar to the first one, this tell a different story. This the graph comparing and finding the difference between the rankings of fast laps with the finishing position on race day. There is a little bias in the nature of the fast lap dataset, as the drivers who have completed all the laps in the race usually have a higher ranking in average pace and usually have a faster lap, which restructures the idea that the data being illsutrated is more than about speed but endurance. Alongside that, the usual factors of chaos and strategy need to be considered here for their rankings, as it takes a whole team to get a result.

First off, the drivers on the line (Kimi Antonelli, Alex Albon, and the 2 Caddilac Drivers) show that their race pace was consistent throughout the whole thing even if it was good or bad with an exception for the Cadillac drivers as they both DNFd.

Secondly, the drivers above the line means that they had a net positive result. They either performed better compared to their fast lap ranking. A notable exception is Arvid Lindblad, as he was incredibly slow, being ranked 20th on fast laps, yet he managed to have net positive 10 by having a P10 result.

Thirdly, the drivers bellow the line illustrate they could've performed better with their result. The Ferraris were the two fastest cars yet underperformed in converting pace into result. An outlier in this side of the graph is once again Piastri, which happened due to a DNF.

Finally, I would like to note a unique observation about this graph, as this graph elaborates the fact that Formula One, through and through is a team sport. Some teams like Ferrari and Red Bull, both drivers are on the same side of the graph, which illustrates the general trend of their performance either due to their speed that weekend or the car just improving. While some teams like Mercedes and Haas have drivers on different parts of the graph, and it shows any gaps of performance between teammates.
